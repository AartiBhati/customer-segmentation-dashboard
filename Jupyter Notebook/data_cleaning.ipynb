**Dataset Description**
This Online Retail II data set contains all the transactions occurring for a UK-based and registered, 
non-store online retail between 01/12/2009 and 09/12/2011.
The company mainly sells unique all-occasion gift-ware. Many customers of the company are wholesalers.


**Attribute Information:**

InvoiceNo: Invoice number. Nominal. A 6-digit integral number uniquely assigned to each transaction. If this code starts with the letter 'c', it indicates a cancellation.
StockCode: Product (item) code. Nominal. A 5-digit integral number uniquely assigned to each distinct product.
Description: Product (item) name. Nominal.
Quantity: The quantities of each product (item) per transaction. Numeric.
InvoiceDate: Invice date and time. Numeric. The day and time when a transaction was generated.
UnitPrice: Unit price. Numeric. Product price per unit in sterling (Â£).
CustomerID: Customer number. Nominal. A 5-digit integral number uniquely assigned to each customer.
Country: Country name. Nominal. The name of the country where a customer resides.

```python
# Reset the index
Retail_data.reset_index(drop=True, inplace=True)
Retail_data
```

```python
# Make a dictionary that contain stockCode as key and item_name as value.
StockCode_dict={}
for index, rows in Retail_data.iterrows():
    if rows['StockCode'] in StockCode_dict:
        if StockCode_dict[rows['StockCode']]=='Nan':
            if rows['Description']!='Nan':
                StockCode_dict[rows['StockCode']]=rows['Description']
            else:
                continue
        else:
            continue
    else:
        StockCode_dict[rows['StockCode']]=rows['Description']
print(StockCode_dict)        
    
```

```python
# Let's fill the missing description using the StockCode dictionary
Retail_data.loc[:, 'Description'] = Retail_data['Description'].fillna(
    Retail_data['StockCode'].map(StockCode_dict)
)
```

```python
Retail_data[Retail_data['Description'].isnull()]
```

```python
# Even after using the StockCode, there are still 217 rows with missing description. We don't have any info about them.
# So we will delete these rows from the dataframe.

Retail_data.dropna(subset=['Description'])
```

```python
Retail_data.reset_index(drop=True, inplace=True)
```

```python
# Missing Description data is handled now.
Retail_data
```

```python
InvoiceNo_dict={}
for index, rows in Retail_data.iterrows():
    if rows['InvoiceNo'] in InvoiceNo_dict:
        if InvoiceNo_dict[rows['InvoiceNo']]=='Nan':
            if rows['CutomerID']!='Nan':
                InvoiceNo_dict[rows['InvoiceNo']]=rows['CustomerID']
            else:
                continue
        else:
            continue
    else:
        InvoiceNo_dict[rows['InvoiceNo']]=rows['CustomerID']
print(InvoiceNo_dict)   
```

```python
Retail_data.loc[:, 'CustomerID'] = Retail_data['CustomerID'].fillna(
    Retail_data['InvoiceNo'].map(InvoiceNo_dict)
)
```
